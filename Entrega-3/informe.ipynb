### Introducción



En este informe se analizarán los resultados de las predicciones de ventas de semanas futuras de bebidas de SodAI, obtenidas a través de un _pipeline_ de Airflow que recibe datos nuevos de una semana $t+1$ y predice las ventas para una semana $t+2$.

El pipeline diseñado consiste de 12 tareas, del cual se destacan las siguientes características claves:
- Realiza un entrenamiento de tres modelos en paralelo: `LogisticRegression`, `LGBMClassifier` y `XGBClassifier`. A su vez, cada uno de estos es entrenado utilizando Optuna para optimizar sus hiperparámetros, nuevamente buscando el mejor valor de f1 score en el conjunto de validación. Luego se escoje el modelo con el mejor f1 score en el conjunto de validación para realizar las predicciones. De este entrenamiento, se almacenan los siguientes artefactos del mejor modelo con `mlflow` para su posterior análisis:
    - Valores de métricas en el conjunto de validación: accuracy, precision, recall, f1 score.
    - Gráfico con el historial de optimización
    - Gráfico de coordenadas paralelas de los hiperparámetros
    - Gráfico de importancia de hiperparámetros
    - Gráfico de valores SHAP para las variables más relevantes
- Implementa detección de _drift_ en los datos, para condicionalmente realizar un nuevo entrenamiento en caso de detectar una pérdida de desempeño con el mejor modelo a la fecha. Esta detección se realiza utilizando la predicción de la nueva semana recibida $t+1$ con el último mejor modelo entrenado. En caso de obtener un valor de f1 score inferior al que obtuvo el modelo cuando fue entrenado, se determina que existe una pérdida de desempeño con los nuevos datos, por lo que se gatilla un  entrenamiento de nuevos modelos. Además, se almacena el valor del f1 score que causó el drift como un artefacto de `mlflow`, para posteriores análisis.
> Cabe destacar que en la primera ejecución del pipeline siempre se realiza un entrenamiento, dado que no existe un modelo previo para intentar detectar drift.

### 📌 Análisis individual [3.0 puntos]

#### Preparación

Antes de comenzar el análisis, se importarán librerías y generarán funciones auxiliares para replicar el análisis en cada batch. También se definirá la variable `universe_pairs`, que corresponda a todos los posibles pares de customer_id y product_id, según los datos que se tienen en `productos.parquet` y `clientes.parquet`, con la finalidad de poder determinar cuales se compraron o no, y así poder obtener vectores binarios que permitan calcular métricas de desempeño.

````python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Carga de datos de clientes y productos
clientes_df = pd.read_parquet('aux_data/clientes.parquet')
productos_df = pd.read_parquet('aux_data/productos.parquet')

# Definición de universe_pairs
universe_df = clientes_df.join(productos_df, how='cross')
universe_pairs = set(map(tuple, universe_df[["customer_id", "product_id"]].values))

# Función auxiliar para obtener desempeño por batch
def get_batch_performance(batch_id):
    data_df = pd.read_parquet(f'data/batch_t{batch_id}.parquet')
    data_df = data_df.groupby(by=['customer_id', 'product_id', 'order_id', 'purchase_date']).sum().reset_index()
    predictions_df = pd.read_csv(f'predictions/batch_t{batch_id}.csv', names=['customer_id', 'product_id'])

    true_pairs = set(map(tuple, data_df[["customer_id", "product_id"]].values))
    pred_pairs = set(map(tuple, predictions_df[["customer_id", "product_id"]].values))

    y_true = [pair in true_pairs for pair in universe_pairs]
    y_pred = [pair in pred_pairs for pair in universe_pairs]

    metrics = {
        'accuracy': accuracy_score(y_true, y_pred),
        'precision': precision_score(y_true, y_pred),
        'recall': recall_score(y_true, y_pred),
        'f1': f1_score(y_true, y_pred),
    }

    return metrics
````
